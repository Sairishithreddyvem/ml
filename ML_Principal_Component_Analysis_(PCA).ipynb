```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score
```

```python
df = pd.read_csv('/content/penguins-PCA.csv')

df.head()
```

```python
df.info()
```

```python
# For numerical columns → fill with mean

for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# For categorical columns → fill with most frequent value (mode)

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Convert Text to Numbers (Label Encoding)

le_dict = {}

for col in df.select_dtypes(include='object').columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    le_dict[col] = le
```

```python
df.head()
```

```python
x = df.drop('sex', axis=1)
y = df['sex']
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# 7. PCA (Reduce Dimensions)
# ================================

pca = PCA()

# Try different number of components
param_grid = {'n_components': list(range(1, x.shape[1] + 1))}

grid_pca = GridSearchCV(pca, param_grid, cv=10)
grid_pca.fit(x_scaled)

best_n = grid_pca.best_params_['n_components']
print("Best number of components:", best_n)
```

```python
# Apply PCA
pca = PCA(n_components=best_n)
x_pca = pca.fit_transform(x_scaled)
x_pca
```

```python
# Visualize PCA (First Component Only)
# ================================
plt.figure(figsize=(7,5))
# Plot the first principal component against a dummy axis (e.g., y=0)
# to visualize its distribution, colored by the target variable.
plt.scatter(x_pca[:,0], np.zeros_like(x_pca[:,0]), c=y)
plt.xlabel("PC1")
plt.ylabel("Dummy Axis") # Updated label to reflect 1D visualization
plt.title("PCA Visualization")
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x_pca, y, test_size=0.2, random_state=42)
```

```python
# Decision Tree with Grid Search
# ================================

dt = DecisionTreeClassifier(random_state=42)

param_grid_dt = {
    'criterion': ['gini', 'entropy'],     # <-- added
    'max_depth': [2, 3, 4, 5, 6],
    'min_samples_split': [2, 3, 5, 10],
    'min_samples_leaf': [1, 2, 3, 4]         # optional but useful
}

grid_dt = GridSearchCV(dt, param_grid_dt, cv=10)
grid_dt.fit(x_train, y_train)

best_model = grid_dt.best_estimator_

print("Best parameters:", grid_dt.best_params_)
print("Best criterion:", grid_dt.best_params_['criterion'])
```

```python
y_pred = best_model.predict(x_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average='weighted')

print("Accuracy:", acc)
print("F1 Score:", f1)
```

```python
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```
